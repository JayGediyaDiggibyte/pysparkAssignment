```python
from pyspark.sql.functions import *
```

##### First Task
###### 1. Create a Dataframe as credit_card_df with different read methods 

```python
credit_card_data = [
    ("1234567891234567"), 
    ("5678912345671234"), 
    ("9123456712345678"), 
    ("1234567812341122"), 
    ("1234567812341342") 
]

credit_card_df = spark.createDataFrame(credit_card_data, ["card_number"])
```

```python
# with Show method
credit_card_df.show()
```

```python
# with Display method
display(credit_card_df)
```

##### Second Task
###### 2. Print number of partitions

###### This is method but I am using Severless Cluster so this is not working because RDD API not support to Serverlesss Clusster
###### credit_card_df.rdd.getNumPartitions()

```python
num_of_partition = credit_card_df.select(spark_partition_id())\
    .distinct()\
    .count()

print(f"Number of partitions: {num_of_partition}")
```

##### Third Task
###### 3. Print number of partitions

```python
credit_card_df = credit_card_df.repartition(5)
credit_card_df.explain(True)
```

##### Fourth Task
###### 4. Decrease the partition size back to its original partition size

```python
credit_card_df = credit_card_df.coalesce(1)
credit_card_df.explain(True)
```

##### Fifth Task
###### 5. Create a UDF to print only the last 4 digits marking the remaining digits as * Eg: ************4567 

```python
def mask_card_number(card):
    return "*" * (len(card) - 4) + card[-4:]

mask_udf = udf(mask_card_number)
```

##### Sixed Task
###### 6. output should have 2 columns as card_number, masked_card_number(with output of question 2) 

```python
mask_card_number_df = credit_card_df.withColumn("masked_card_number", mask_udf("card_number"))
mask_card_number_df.display()
mask_card_number_df.explain(True)
```
