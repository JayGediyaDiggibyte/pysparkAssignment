## Fifth Question

```python
from pyspark.sql.functions import *
```

##### First Task
###### 1. create all 3 data frames as employee_df, department_df, country_df with custom schema defined in dynamic way 

```python
emp_data = [
    (11,"james","D101","ny",9000,34), 
    (12,"michel","D101","ny",8900,32), 
    (13,"robert","D102","ca",7900,29), 
    (14,"scott","D103","ca",8000,36), 
    (15,"jen","D102","ny",9500,38), 
    (16,"jeff","D103","uk",9100,35), 
    (17,"maria","D101","ny",7900,40)
]

employee_df = spark.createDataFrame(emp_data, ["employee_id", "employee_name", "department", "state", "salary", "age"])

display(employee_df)
```

```python
dept_data = [
    ("D101","sales"), 
    ("D102","finance"), 
    ("D103","marketing"), 
    ("D104","hr"), 
    ("D105","support")
]

department_df = spark.createDataFrame(dept_data, ["dept_id", "dept_name"])
display(department_df)
```

```python
country_data = [
    ("ny","newyork"), 
    ("ca","California"), 
    ("uk","Russia")
]

country_df = spark.createDataFrame(country_data, ["country_code", "country_name"])
display(country_df)
```

##### Second Task
###### 2. Find avg salary of each department 

```python
avg_salary_df = employee_df.groupBy("department").agg(avg("salary").alias("avg_salary"))
display(avg_salary_df)
```

##### Third Task
###### 3. Find the employee’s name and department name whose name starts with ‘m’ 

```python
m_emp_df = employee_df.join(department_df, employee_df.department == department_df.dept_id, "inner")\
                    .filter(col("employee_name").startswith("m"))\
                    .select("employee_name", "dept_name")

display(m_emp_df)
```

##### Fourth Task
###### 4. Create another new column in  employee_df as a bonus by multiplying employee salary *2

```python
employee_df = employee_df.withColumn("bonus_salary", col("salary") * 2)
display(employee_df)
```

##### Fifth Task
###### 5. Reorder the column names of employee_df columns as (employee_id,employee_name,salary,State,Age,department) 

```python
employee_df = employee_df.select("employee_id", "employee_name", "salary", "state", "age", "department")
display(employee_df)
```

##### Sixth Task
###### 6. Give the result of an inner join, left join, and right join when joining employee_df with department_df in a dynamic way  

```python
# Inter Join

inner_join_df = employee_df.join(department_df, employee_df.department == department_df.dept_id, "inner")
display(inner_join_df)
```

```python
# Left Join

left_join_df = employee_df.join(department_df, employee_df.department == department_df.dept_id, "left")
display(left_join_df)
```

```python
# Right Join

right_join_df = employee_df.join(department_df, employee_df.department == department_df.dept_id, "right")
display(right_join_df)
```

##### Seventh Task
###### 7. Derive a new data frame with country_name instead of State in employee_df  Eg(11,“james”,”D101”,”newyork”,8900,32) 

```python
country_employee_df = employee_df.join(country_df, employee_df.state == country_df.country_code, "left")\
    .select("employee_id", "employee_name", "department", col("country_name").alias("state"), "salary")

display(country_employee_df)
```

##### Eight Task
###### 8. convert all the column names into lowercase from the result of question 7in a dynamic way, add the load_date column with the current date 

```python
new_cols = [c.lower() for c in country_employee_df.columns]

country_employee_df = country_employee_df.toDF(*new_cols)

country_employee_df = country_employee_df.withColumn("current_date", current_date())

display(country_employee_df)
```

##### Ninth Task
###### 9. create 2 external tables with parquet, CSV format with the same name database name, and 2 different table names as CSV and parquet format. 

```python
spark.sql("CREATE DATABASE IF NOT EXISTS employee")
```

```python

country_employee_df.write \
.mode("overwrite") \
.format("parquet") \
.saveAsTable("employee.employee_parquet")
```

```python

country_employee_df.write \
.mode("overwrite") \
.option("header","true") \
.format("csv") \
.saveAsTable("employee.employee_csv")
```
