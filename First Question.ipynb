## First Question

```python
from pyspark.sql.functions import *
```

##### First Task
###### 1. Create DataFrame as purchase_data_df,  product_data_df with custom schema with the below data 

```python
purchase_data = [
    (1, "iphone13"), 
    (1, "dell i5 core"), 
    (2, "iphone13"), 
    (2, "dell i5 core"), 
    (3, "iphone13"), 
    (3, "dell i5 core"), 
    (1, "dell i3 core"), 
    (1, "hp i5 core"), 
    (1, "iphone14"), 
    (3, "iphone14"), 
    (4, "iphone13") 
]

purchase_data_df = spark.createDataFrame(
    purchase_data, 
    ["customer_id", "product_model"]
)
```

```python
product_model = [
    ("iphone13"), 
    ("dell i5 core"), 
    ("dell i3 core"), 
    ("hp i5 core"), 
    ("iphone14") 
]

product_data_df = spark.createDataFrame(
    product_model, 
    ["product_model"]
)
```

```python
purchase_data_df.display()
product_data_df.display()
```

##### Second Task
###### 2. Find the customers who have bought only iphone13

```python
iphone13_df = purchase_data_df.filter(col("product_model") == "iphone13")
display(iphone13_df)
```

##### Third Task
###### 3. Find customers who upgraded from product iphone13 to product iphone14 

```python
upgrade_df = purchase_data_df.groupBy("customer_id").agg(collect_set("product_model").alias("products")) \
    .filter(array_contains(col("products"), "iphone13") & array_contains(col("products"), "iphone14"))
display(upgrade_df)
```

##### Fourth Task
###### 4. Find customers who have bought all models in the new Product Data

```python
total_product = product_data_df.count()

all_product_buy_df = purchase_data_df.join(product_data_df, "product_model")\
                .groupBy("customer_id")\
                    .agg(countDistinct("product_model").alias("total_product"))\
                        .filter(col("total_product") == total_product)
display(all_product_buy_df)
```
